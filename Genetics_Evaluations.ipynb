```python
import random

def generate_tomato_chromosome():
    """
    Generates a single chromosome representing a tomato plant.
    The chromosome is an array of 3 genes: [size, color, sweetness].
    """
    # Gene values range from 0 to 5
    size = random.randint(0, 5)       # 0 = Small  -> 5 = Large
    color = random.randint(0, 5)      # 0 = Green  -> 5 = Red
    sweetness = random.randint(0, 5)  # 0 = Sour   -> 5 = Sweet

    return [size, color, sweetness]

def decode_traits(chromosome):
    """
    Translates the numerical genes into readable physical traits.
    """
    size, color, sweetness = chromosome

    # Decode Size
    size_desc = "Small" if size <= 2 else "Large"

    # Decode Color
    color_desc = "Green" if color <= 2 else "Red"

    # Decode Sweetness
    sweet_desc = "Sour" if sweetness <= 2 else "Sweet"

    return f"Size: {size_desc} ({size}), Color: {color_desc} ({color}), Taste: {sweet_desc} ({sweetness})"

# --- Main Execution ---

# 1. Initialize two parent chromosomes
parent_1 = generate_tomato_chromosome()
parent_2 = generate_tomato_chromosome()

# 2. Display the genetic makeup and physical traits
print("--- Tomato Parent 1 ---")
print("Genes (Chromosome):", parent_1)
print("Physical Traits:", decode_traits(parent_1))

print("\n--- Tomato Parent 2 ---")
print("Genes (Chromosome):", parent_2)
print("Physical Traits:", decode_traits(parent_2))
```
